**Q1. What is Min-Max scaling, and how is it used in data preprocessing? Provide an example to illustrate its 
application.**

Min-Max scaling is a data preprocessing technique used to transform numerical features into a specific range, typically between 0 and 1. It's employed to handle features with different scales and units, ensuring they contribute equally to the model.

For Instance :
Identify the minimum (min) and maximum (max) values in the dataset for the feature.
Calculate the scaled value for each data point using the formula:
scaled_value = (value - min) / (max - min)
Multiply the scaled value by the desired range (e.g., 1 for 0-1 scaling) and add the minimum of the desired range.
Example:
Suppose we have a dataset of house prices ranging from $100,000 to $500,000. To scale prices to a range of 0 to 1:

**Q2. What is the Unit Vector technique in feature scaling, and how does it differ from Min-Max scaling? 
Provide an example to illustrate its application..**

Unit vector technique is another feature scaling method, but it's different from Min-Max scaling. It involves dividing each feature by its Euclidean norm, resulting in vectors of length 1.

How it works:

Calculate the Euclidean norm of each feature.
Divide each data point in the feature by the calculated norm.
Example:
Suppose we have two features: height and weight. To apply unit vector scaling:

Calculate the Euclidean norm for each data point based on height and weight.
Divide each height and weight value by the corresponding norm.
Difference from Min-Max scaling:

Min-Max scaling preserves the original distribution of the data while scaling it to a specific range.
Unit vector scaling changes the distribution of the data by making all data points lie on a unit sphere.


**Q3. What is PCA (Principle Component Analysis), and how is it used in dimensionality reduction? Provide an 
example to illustrate its application.**

Principal Component Analysis (PCA) is a dimensionality reduction technique that transforms a large set of variables into a smaller set that still contains most of the information in the original set. It does so by identifying the directions (principal components) along which the variance of the data is maximized.

Steps in PCA:

Standardize the data.
Compute the covariance matrix of the data.
Calculate the eigenvalues and eigenvectors of the covariance matrix.
Sort the eigenvalues and corresponding eigenvectors in descending order.
Select the top k eigenvectors to form a new feature space.
Transform the original data into the new feature space.
Example:

Consider a dataset with two features: height and weight. After standardizing the data and computing the covariance matrix, we obtain the eigenvalues and eigenvectors. Suppose the top two eigenvalues correspond to eigenvectors [0.6, 0.8] and [-0.8, 0.6]. We can project the original data onto these eigenvectors to get a reduced dataset.

**Q4. What is the relationship between PCA and Feature Extraction, and how can PCA be used for Feature 
Extraction? Provide an example to illustrate this concept.**

Feature Extraction involves creating new features from the original features that capture the most important information. PCA can be used for feature extraction by selecting the principal components as the new features.

Example:

In a dataset with features such as height, weight, and age, applying PCA might reveal that the first principal component (PC1) captures most of the variance and could be interpreted as a combination of height and weight. The second principal component (PC2) might capture the remaining variance and could be related to age. By using PC1 and PC2 as new features, we can reduce the dimensionality of the dataset while retaining most of the information.
